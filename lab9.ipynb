```python

def eliminate_implication(expr):
    expr = expr.replace(" -> ", " OR NOT ")
    return expr


def move_negation(expr):
    expr = expr.replace("NOT NOT ", "")
    return expr


def distribute_or(expr):
    # Example: A OR (B AND C)
    if " OR (" in expr and " AND " in expr:
        parts = expr.split(" OR ")
        a = parts[0]
        b = parts[1].strip("()")
        terms = b.split(" AND ")

        result = []
        for term in terms:
            result.append("(" + a + " OR " + term + ")")

        return " AND ".join(result)

    return expr


def convert_to_cnf(expr):
    print("Original Expression:", expr)

    expr = eliminate_implication(expr)
    print("After eliminating implication:", expr)

    expr = move_negation(expr)
    print("After moving negation:", expr)

    expr = distribute_or(expr)
    print("CNF Expression:", expr)

    return expr


# Main program
expression = input("Enter FOL expression: ")
convert_to_cnf(expression)
```
