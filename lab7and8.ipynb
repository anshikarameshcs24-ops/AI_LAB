```python
def unify(x, y):
  if x == y:
    return {}

  if x.islower():
    if x in y:
      return "FAILURE"
      return {x: y}

  if y.islower():
    if y in x:
      return "FAILURE"
      return {y: x}
      return "FAILURE"



x = input("Enter first expression: ")
y = input("Enter second expression: ")

result = unify(x, y)

if result != "FAILURE":
  print("Unification Successful!")
  print("Substitution:", result)
else:
  print("Unification Failed!")
```

```python
# Initial Knowledge Base (Facts)
KB = [
    ("American", "Robert"),
    ("Weapon", "T1"),
    ("Sells", "Robert", "T1", "A"),
    ("Hostile", "A")
]

# Forward Chaining Algorithm
def forward_chaining(KB):

    new_fact = True

    while new_fact:
        new_fact = False

        # Check the rule:
        # American(p) AND Weapon(q) AND Sells(p,q,r) AND Hostile(r)
        #                => Criminal(p)

        for a in KB:
            for w in KB:
                for s in KB:
                    for h in KB:

                        if (a[0] == "American" and
                            w[0] == "Weapon" and
                            s[0] == "Sells" and
                            h[0] == "Hostile"):

                            if (a[1] == s[1] and
                                w[1] == s[2] and
                                h[1] == s[3]):

                                fact = ("Criminal", a[1])

                                if fact not in KB:
                                    KB.append(fact)
                                    new_fact = True

    return KB


# Execute algorithm
result = forward_chaining(KB)

print("Knowledge Base after Forward Chaining:")
for fact in result:
    print(fact)

# Check query
if ("Criminal", "Robert") in result:
    print("\nRobert is a Criminal")
else:
    print("\nRobert is not a Criminal")
```

```python
KB = {
    "Barks(Tommy)",
    "FourLegs(Tommy)",
    "SharpTeeth(Tommy)"
}

print("Initial Facts:")
for fact in KB:
    print(fact)

print("\nForward Chaining:")

while True:
    new_facts = set()

    if "Barks(Tommy)" in KB and "FourLegs(Tommy)" in KB:
        new_facts.add("Dog(Tommy)")

    if "Feathers(Tommy)" in KB and "LaysEggs(Tommy)" in KB:
        new_facts.add("Bird(Tommy)")

    if "Dog(Tommy)" in KB:
        new_facts.add("Mammal(Tommy)")

    if "Mammal(Tommy)" in KB and "SharpTeeth(Tommy)" in KB:
        new_facts.add("Carnivore(Tommy)")

    if "Bird(Tommy)" in KB:
        new_facts.add("CanFly(Tommy)")

    new_facts = new_facts - KB

    if not new_facts:
        break

    for fact in new_facts:
        KB.add(fact)
        print("Derived:", fact)

print("\nFinal Result:")

if "Carnivore(Tommy)" in KB:
    print("Goal Proved!")
    print("Tommy is a Carnivore.")
else:
    print("Goal Not Proved!")
```
